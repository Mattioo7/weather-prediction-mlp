## 1. Inspekcja struktury danych wejściowych

Celem tego notebooka jest wczytanie plików CSV z katalogu `data/` oraz
wypisanie nazw kolumn dla każdego pliku, aby jednoznacznie określić
strukturę danych wejściowych wykorzystywanych w projekcie.

Struktura katalogów:

- `data/city_attributes.csv`
- `data/train/` – zbiór danych właściwy (wszystkie pliki CSV)
- `data/test/` – zbiór o identycznej strukturze (nieużywany na tym etapie)


```python
from pathlib import Path
import pandas as pd
```

```python
# notebooks/ -> project_root/
PROJECT_ROOT = Path("..").resolve()

DATA_DIR = PROJECT_ROOT / "data"
CITY_ATTR_FILE = DATA_DIR / "city_attributes.csv"
TRAIN_DIR = DATA_DIR / "train"
```

```python
print(f"DATA DIRECTORY: {DATA_DIR.resolve()}")
print(f"CITY ATTRIBUTES FILE: {CITY_ATTR_FILE.resolve()}")
print(f"TRAIN DIRECTORY: {TRAIN_DIR.resolve()}")
```

```python
def print_csv_columns(csv_path: Path):
    df = pd.read_csv(csv_path, nrows=0)  # tylko nagłówki
    print(f"\n{csv_path}")
    print("-" * len(str(csv_path)))
    for col in df.columns:
        print(col)
```

```python
print("CITY ATTRIBUTES")
print("=" * 20)
print_csv_columns(CITY_ATTR_FILE)
```

```python
print("\nTRAIN FILES")
print("=" * 20)

for csv_file in sorted(TRAIN_DIR.glob("*.csv")):
    print_csv_columns(csv_file)
```

## 2. Analiza zmiennej jakościowej `weather_description`

Plik `weather_description_train.csv` zawiera tekstowe opisy warunków pogodowych
dla poszczególnych miast i momentów czasowych.

Celem tej sekcji jest określenie liczby unikalnych opisów pogody.


```python
from pathlib import Path
import pandas as pd

# notebooks/ -> project_root/
PROJECT_ROOT = Path("..").resolve()

DATA_DIR = PROJECT_ROOT / "data"
TRAIN_DIR = DATA_DIR / "train"
WEATHER_DESCRIPTION_FILE = TRAIN_DIR / "weather_description_train.csv"

print(f"DATA DIRECTORY: {DATA_DIR.resolve()}")
print(f"TRAIN DIRECTORY: {TRAIN_DIR.resolve()}")
print(f"WEATHER DESCRIPTION FILE: {WEATHER_DESCRIPTION_FILE.resolve()}")
```

```python
weather_desc_df = pd.read_csv(WEATHER_DESCRIPTION_FILE, sep=";")
weather_desc_df.head()
```

```python
description_values = weather_desc_df.drop(columns=["datetime"]).values.flatten()
description_series = pd.Series(description_values).dropna()

print(f"Liczba obserwacji: {len(description_series)}")
print(f"Liczba unikalnych opisów pogody: {description_series.nunique()}")
```

```python
description_counts = description_series.value_counts()

description_counts.head(10)
```

```python
description_freq = (description_counts / description_counts.sum()) * 100

description_freq.head(10)
```

```python
rare_descriptions = description_counts[description_counts < 100]

print(f"Liczba rzadkich opisów (<100 wystąpień): {len(rare_descriptions)}")
rare_descriptions.sort_values().head(30)
```
